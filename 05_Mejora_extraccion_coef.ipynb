```python
# Function to extract each coefficient of CCDC. In this part just coeficients for the firsty three timebreaks 
def extract_bands(direct,folder):
        

        
    B_t1=gdal.Open(direct+'\\'+folder[0]) 
    B_t2=gdal.Open(direct+'\\'+folder[1])
    B_t3=gdal.Open(direct+'\\'+folder[2])
    B_t4=gdal.Open(direct+'\\'+folder[3])
    B_t5=gdal.Open(direct+'\\'+folder[4])
    B_t6=gdal.Open(direct+'\\'+folder[5])
    B_t7=gdal.Open(direct+'\\'+folder[6])
    B_t8=gdal.Open(direct+'\\'+folder[7])
    #Extract the first band... in this case coef of first segment 

    B1_t1=B_t1.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t2=B_t2.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t3=B_t3.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t4=B_t4.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t5=B_t5.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t6=B_t6.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t7=B_t7.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B1_t8=B_t8.GetRasterBand(1).ReadAsArray().astype(np.float32)


    #Extract the second band... in this case  coef of second segment 
    B2_t1=B_t1.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t2=B_t2.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t3=B_t3.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t4=B_t4.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t5=B_t5.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t6=B_t6.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t7=B_t7.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B2_t8=B_t8.GetRasterBand(2).ReadAsArray().astype(np.float32)

    # Extract the third band... in this case  coef of 3rd segment
    B3_t1=B_t1.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t2=B_t2.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t3=B_t3.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t4=B_t4.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t5=B_t5.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t6=B_t6.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t7=B_t7.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B3_t8=B_t8.GetRasterBand(3).ReadAsArray().astype(np.float32)

    return B1_t1,B1_t2,B1_t3,B1_t4,B1_t5,B1_t6,B1_t7,B1_t8,B2_t1,B2_t2,B2_t3,B2_t4,B2_t5,B2_t6,B2_t7,B2_t8,B3_t1,B3_t2,B3_t3,B3_t4,B3_t5,B3_t6,B3_t7,B3_t8
```

```python
# Function to extract each coefficient of CCDC. In this part just coeficients from the 4th to 7th timebreaks 

def extract_bands_4_7(direct,folder):
        

        
    B_t1=gdal.Open(direct+'\\'+folder[0]) 
    B_t2=gdal.Open(direct+'\\'+folder[1])
    B_t3=gdal.Open(direct+'\\'+folder[2])
    B_t4=gdal.Open(direct+'\\'+folder[3])
    B_t5=gdal.Open(direct+'\\'+folder[4])
    B_t6=gdal.Open(direct+'\\'+folder[5])
    B_t7=gdal.Open(direct+'\\'+folder[6])
    B_t8=gdal.Open(direct+'\\'+folder[7])
    #Extract the first band... in this case coef of first segment 

    B4_t1=B_t1.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t2=B_t2.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t3=B_t3.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t4=B_t4.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t5=B_t5.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t6=B_t6.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t7=B_t7.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B4_t8=B_t8.GetRasterBand(1).ReadAsArray().astype(np.float32)


    #Extract the second band... in this case  coef of second segment 
    B5_t1=B_t1.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t2=B_t2.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t3=B_t3.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t4=B_t4.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t5=B_t5.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t6=B_t6.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t7=B_t7.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B5_t8=B_t8.GetRasterBand(2).ReadAsArray().astype(np.float32)

    # Extract the third band... in this case  coef of 3rd segment
    B6_t1=B_t1.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t2=B_t2.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t3=B_t3.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t4=B_t4.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t5=B_t5.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t6=B_t6.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t7=B_t7.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B6_t8=B_t8.GetRasterBand(3).ReadAsArray().astype(np.float32)

    return B4_t1,B4_t2,B4_t3,B4_t4,B4_t5,B4_t6,B4_t7,B4_t8,B5_t1,B5_t2,B5_t3,B5_t4,B5_t5,B5_t6,B5_t7,B5_t8,B6_t1,B6_t2,B6_t3,B6_t4,B6_t5,B6_t6,B6_t7,B6_t8
```

```python
#Function to get for each pixel the value of each coeficient

def upload_shp(name, bands,shp):
    val=[]
    rows=list(shp['Rows'])
    col=list(shp['Columns'])
    for i in range (len(rows)):
        extract=bands[rows[i]][col[i]]
        val.append(extract)
    
    shp[name]=val
        
    return shp
```

```python
# Function to extract tEnd and tStart Navarre

def extract_time(direct,folder):
                
    B_t1=gdal.Open(direct+'\\'+folder[0]) 
    B_t2=gdal.Open(direct+'\\'+folder[1])
    B_t3=gdal.Open(direct+'\\'+folder[2])
    B_t4=gdal.Open(direct+'\\'+folder[3])
    B_t5=gdal.Open(direct+'\\'+folder[4])
    B_t6=gdal.Open(direct+'\\'+folder[5])
    B_tS=gdal.Open(direct+'\\'+folder[6]) 
    
    #Extract the first band... in this case time end

    t1=B_t1.GetRasterBand(1).ReadAsArray().astype(np.float32)
    t2=B_t2.GetRasterBand(1).ReadAsArray().astype(np.float32)
    t3=B_t3.GetRasterBand(1).ReadAsArray().astype(np.float32)
    t4=B_t4.GetRasterBand(1).ReadAsArray().astype(np.float32)
    t5=B_t5.GetRasterBand(1).ReadAsArray().astype(np.float32)
    t6=B_t6.GetRasterBand(1).ReadAsArray().astype(np.float32)
    
    # Time start
    B_tS1=B_tS.GetRasterBand(1).ReadAsArray().astype(np.float32)
    B_tS2=B_tS.GetRasterBand(2).ReadAsArray().astype(np.float32)
    B_tS3=B_tS.GetRasterBand(3).ReadAsArray().astype(np.float32)
    B_tS4=B_tS.GetRasterBand(4).ReadAsArray().astype(np.float32)
    B_tS5=B_tS.GetRasterBand(5).ReadAsArray().astype(np.float32)
    B_tS6=B_tS.GetRasterBand(6).ReadAsArray().astype(np.float32)
  
    return t1,t2,t3,t4,t5,t6,B_tS1,B_tS2,B_tS3,B_tS4,B_tS5,B_tS6
```

```python
# Detection of the peak before the disturbance
def detect_peaks(valores,x_s,x_f, min_height=0.01):
    principio=int(x_s.min())

    fin=int(x_f.min())
    x=fin-principio
    if x<1:
        peaks = argrelextrema(valores, np.greater, order=4)[0]
    else:
        dist=((371/x)/2)-2
        distance=int(dist)
        print(distance)
        peaks = argrelextrema(valores, np.greater, order=distance)[0]
    return [peak for peak in peaks if valores[peak] >= min_height]

def detect_peaks_last(valores,min_height=0.01):

    peaks = argrelextrema(valores, np.greater, order=4)[0]
    
    return [peak for peak in peaks if valores[peak] >= min_height]


# Calculate coefficients to recreate the CCDC fit
def coef(select):
    # Before
    coefficients_1=np.array([select['Coef_1_1'],select['Coef_1_2'],select['Coef_1_3'],select['Coef_1_4'],select['Coef_1_5'],select['Coef_1_6'],select['Coef_1_7'],select['Coef_1_8']])
    # After
    coefficients_2=np.array([select['Coef_2_1'],select['Coef_2_2'],select['Coef_2_3'],select['Coef_2_4'],select['Coef_2_5'],select['Coef_2_6'],select['Coef_2_7'],select['Coef_2_8']])

    # print('Coeficientes: ',coefficients_1[1])
    return coefficients_1,coefficients_2

# Consider the start date and split the time serie in order to have the same frequency in the time serie.
def x_axes(select):
    # Before
    x_1 = np.linspace(select['Ts1'], select['t1'], int((1 - 0) /0.0027) + 1)
    # After
    x_2=np.linspace(select['Ts2'], select['t2'], int((1 - 0) / 0.0027) + 1)

    # print('Eje x: ',x_1)
    return x_1,x_2

#Calculate the coefficients based on the function developed before and fit the model after and before the fire.

def trigger(data):
    coefficients_1,coefficients_2=coef(data)
    x_1,x_2=x_axes(data)
    y_pred=fit_function(x_1,coefficients_1)
    y_after=fit_function(x_2,coefficients_2)
    # print('Y predicho: ',y_pred)
    return y_pred,y_after,x_1,x_2

# Fit the function considering the different coefficients
def reemplazar_valores(array):
    return np.array([1 if abs(x) > 1 else x for x in array])



def fit_function(x, coefficients):
    # Extract individual coefficients
    intercept = coefficients[0]
    slope = coefficients[1]
    sin_coeff = coefficients[2]
    cos_coeff = coefficients[3]
    sin2_coeff = coefficients[4]
    cos2_coeff = coefficients[5]
    sin3_coeff = coefficients[6]
    cos3_coeff = coefficients[7]

    # Calculate the predicted values using the coefficients and input data. Formula based on formula used in Google Earth Engine
    pi=math.pi
    Pi2=(2*pi)
    #Pi2=pi
    
    y_pred=intercept + slope * x  + sin_coeff * np.sin(x*Pi2) + cos_coeff * np.cos(x*Pi2) + sin2_coeff * np.sin(2 *x*Pi2) + cos2_coeff * np.cos(2 *x*Pi2)+ sin3_coeff * np.sin(3 *x*Pi2) + cos3_coeff * np.cos(3 *x*Pi2)

    y_pred=reemplazar_valores(y_pred)
    #y_pred=intercept + slope * x  + sin_coeff *x* np.sin(Pi2) + cos_coeff *x* np.cos(Pi2) + sin2_coeff *x* np.sin(2 *Pi2) + cos2_coeff *x* np.cos(2 *Pi2)+ sin3_coeff *x* np.sin(3 *Pi2) + cos3_coeff *x* np.cos(3 *Pi2)
    
    return y_pred
```

```python
def coef_sig(select):
    # Before
    coefficients_3=np.array([select['Coef_3_1'],select['Coef_3_2'],select['Coef_3_3'],select['Coef_3_4'],select['Coef_3_5'],select['Coef_3_6'],select['Coef_3_7'],select['Coef_3_8']])
    # After
    coefficients_4=np.array([select['Coef_4_1'],select['Coef_4_2'],select['Coef_4_3'],select['Coef_4_4'],select['Coef_4_5'],select['Coef_4_6'],select['Coef_4_7'],select['Coef_4_8']])
    
    coefficients_5=np.array([select['Coef_5_1'],select['Coef_5_2'],select['Coef_5_3'],select['Coef_5_4'],select['Coef_5_5'],select['Coef_5_6'],select['Coef_5_7'],select['Coef_5_8']])
    
    coefficients_6=np.array([select['Coef_6_1'],select['Coef_6_2'],select['Coef_6_3'],select['Coef_6_4'],select['Coef_6_5'],select['Coef_6_6'],select['Coef_6_7'],select['Coef_6_8']])
    

    return coefficients_3,coefficients_4,coefficients_5,coefficients_6

# Consider the start date and split the time serie in order to have the same frequency in the time serie.
def x_axes_sig(select):
    # Before
    x_3 = np.linspace(select['Ts3'], select['t3'], int((1 - 0) /0.0027) + 1)
    # After
    x_4=np.linspace(select['Ts4'], select['t4'], int((1 - 0) / 0.0027) + 1)
    
    x_5=np.linspace(select['Ts5'], select['t5'], int((1 - 0) / 0.0027) + 1)
    
    x_6=np.linspace(select['Ts6'], select['t6'], int((1 - 0) / 0.0027) + 1)

    return x_3,x_4,x_5,x_6

#Calculate the coefficients based on the function developed before and fit the model after and before the fire.

def trigger_sig(data):
    coefficients_3,coefficients_4,coefficients_5,coefficients_6=coef_sig(data)
    x_3,x_4,x_5,x_6=x_axes_sig(data)
    y_3=fit_function(x_3,coefficients_3)
    y_4=fit_function(x_4,coefficients_4)
    y_5=fit_function(x_5,coefficients_5)
    y_6=fit_function(x_6,coefficients_6)

    return y_3,y_4,y_5,y_6,x_3,x_4,x_5,x_6
```

```python
def found_peak(data, threshold=0.01):
  """
  Encuentra los picos en un array de datos.

  Args:
    data: Un array de valores numéricos.
    threshold: Un valor umbral para la amplitud del pico.

  Returns:
    Un array de índices correspondientes a los picos encontrados.
  """

  peaks = []
  for i in range(1, len(data) - 1):
    if data[i] > data[i-1] and data[i] > data[i+1] and data[i] > threshold:
      peaks.append(i)
  return peaks
```
